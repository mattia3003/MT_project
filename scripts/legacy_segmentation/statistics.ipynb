```python
corruption_file = Path("../docs/corrupted_synthetic_masks_10_0.txt")

records = []
for line in corruption_file.read_text(encoding="utf-8").splitlines():
    line = line.strip()
    if not line or line.startswith("#"):
        continue

    filename, reason = [part.strip() for part in line.split("|", maxsplit=1)]
    outside_match = re.search(r"Pixels outside brain:\s*(\d+)", reason)
    blobs_match = re.search(r"Excessive noise fragments:\s*(\d+)", reason)

    records.append(
        {
            "filename": filename,
            "outside_brain_pixels": (
                int(outside_match.group(1)) if outside_match else None
            ),
            "distinct_blobs": (
                int(blobs_match.group(1)) if blobs_match else None
            ),
        }
    )

corruptions = pd.DataFrame(records)

# Confirm that the report has the same number of files as the statistics table.
assert len(corruptions) == 2243
assert corruptions["filename"].is_unique

blob_counts = corruptions["distinct_blobs"].dropna()
lone_pixel_count = int(
    (
        corruptions["outside_brain_pixels"].eq(1)
        & corruptions["distinct_blobs"].isna()
    ).sum()
)

base_statistics = {
    "Corrupt data": ["Synthetic"],
    "Total count": [len(corruptions)],
    "Percentage of total slices": [round(len(corruptions) / 37624 * 100, 2)],
    "Lone pixel count": [lone_pixel_count],
}

# Cumulative thresholds: a file with 120 blobs appears in every applicable column.
synthetic_dataset_statistics = pd.DataFrame(
    {
        **base_statistics,
        **{
            f"> {threshold} blobs": [int(blob_counts.gt(threshold).sum())]
            for threshold in (10, 20, 50, 100)
        },
    }
)

# Mutually exclusive ranges: 10-20 means 10 <= blobs < 20, and so on.
synthetic_dataset_range_statistics = pd.DataFrame(
    {
        **base_statistics,
        "10-20 blobs": [int(((blob_counts >= 10) & (blob_counts < 20)).sum())],
        "20-50 blobs": [int(((blob_counts >= 20) & (blob_counts < 50)).sum())],
        "50-100 blobs": [int(((blob_counts >= 50) & (blob_counts <= 100)).sum())],
        "> 100 blobs": [int((blob_counts > 100).sum())],
    }
)

assert (
    synthetic_dataset_range_statistics[
        ["10-20 blobs", "20-50 blobs", "50-100 blobs", "> 100 blobs"]
    ].sum(axis=1).iloc[0]
    == synthetic_dataset_statistics["> 10 blobs"].iloc[0]
)

print("Synthetic dataset statistics:")
display(
    HTML(
        "<div style='display:flex; gap:24px; align-items:flex-start; overflow-x:auto;'>"
        "<div><h4>Cumulative thresholds</h4>"
        + synthetic_dataset_statistics.to_html(index=False, border=1)
        + "</div><div><h4>Exclusive ranges</h4>"
        + synthetic_dataset_range_statistics.to_html(index=False, border=1)
        + "</div></div>"
    )
)
```

#### 1.1.1 Synthetic mask geometry and image statistics before and after filtering

This analysis compares the complete synthetic dataset with the conservative filtered dataset. The filtered set is defined by the existing exclusion list and corresponding converted directory. Since no independent brain mask is available, low-intensity pixels are reported descriptively and are not interpreted as anatomical brain-background pixels.

The measures include mask area, bounding-box dimensions and fill ratio, connected-component counts using 4- and 8-connectivity, image intensity statistics, negative-pixel fraction, low-intensity fraction, and intensity statistics inside the metastasis mask.

```python
from scipy.ndimage import label, generate_binary_structure

SYNTHETIC_ALL_DIR = Path("../processed_data/Synthetic_data_2d_all")
SYNTHETIC_FILTERED_DIR = Path("../processed_data/Synthetic_data_2d_10_0")
LOW_INTENSITY_THRESHOLD = 0.01

def _paired_synthetic_files(root):
    bravo = {p.name.removesuffix(".bravo.nii.gz"): p for p in (root / "bravo").glob("*.nii*")}
    seg = {p.name.removesuffix(".seg.nii.gz"): p for p in (root / "seg").glob("*.nii*")}
    if set(bravo) != set(seg):
        raise ValueError(f"BRAVO/SEG mismatch in {root}: {len(bravo)} images, {len(seg)} masks")
    return [(key, bravo[key], seg[key]) for key in sorted(bravo)]

def _summarize_synthetic_dataset(root, dataset_name):
    rows = []
    for stem, image_path, mask_path in _paired_synthetic_files(root):
        image = np.asarray(nib.load(str(image_path)).dataobj).squeeze().astype(np.float32) # type: ignore
        mask = np.asarray(nib.load(str(mask_path)).dataobj).squeeze() > 0.5 # type: ignore
        if image.shape != mask.shape:
            raise ValueError(f"Shape mismatch for {stem}: {image.shape} vs {mask.shape}")
        positive = np.argwhere(mask)
        area = int(mask.sum())
        if area:
            y0, x0 = positive.min(axis=0)
            y1, x1 = positive.max(axis=0)
            bbox_height, bbox_width = int(y1 - y0 + 1), int(x1 - x0 + 1)
            bbox_area = bbox_height * bbox_width
            fill_ratio = area / bbox_area
        else:
            bbox_height = bbox_width = bbox_area = 0
            fill_ratio = np.nan
        components_4 = int(label(mask, structure=generate_binary_structure(2, 1))[1]) # type: ignore
        components_8 = int(label(mask, structure=generate_binary_structure(2, 2))[1]) # type: ignore
        masked_values = image[mask]
        rows.append({
            "dataset": dataset_name, "stem": stem, "mask_area_px": area,
            "bbox_height_px": bbox_height, "bbox_width_px": bbox_width,
            "bbox_area_px": bbox_area, "mask_fill_ratio": fill_ratio,
            "components_4": components_4, "components_8": components_8,
            "image_min": float(image.min()), "image_max": float(image.max()),
            "image_mean": float(image.mean()), "image_std": float(image.std()),
            "image_p01": float(np.percentile(image, 1)),
            "image_median": float(np.median(image)),
            "image_p99": float(np.percentile(image, 99)),
            "negative_fraction": float((image < 0).mean()),
            "low_intensity_fraction": float((image <= LOW_INTENSITY_THRESHOLD).mean()),
            "mask_image_mean": float(masked_values.mean()) if area else np.nan,
            "mask_image_std": float(masked_values.std()) if area else np.nan,
            "mask_image_min": float(masked_values.min()) if area else np.nan,
            "mask_image_max": float(masked_values.max()) if area else np.nan,
            "mask_low_intensity_fraction": float((masked_values <= LOW_INTENSITY_THRESHOLD).mean()) if area else np.nan,
        })
    return pd.DataFrame(rows)

synthetic_geometry_all = _summarize_synthetic_dataset(SYNTHETIC_ALL_DIR, "Synthetic all")
synthetic_geometry_filtered = _summarize_synthetic_dataset(SYNTHETIC_FILTERED_DIR, "Synthetic filtered")
synthetic_geometry = pd.concat([synthetic_geometry_all, synthetic_geometry_filtered], ignore_index=True)

geometry_columns = ["mask_area_px", "bbox_height_px", "bbox_width_px", "bbox_area_px", "mask_fill_ratio", "components_4", "components_8"]
image_columns = ["image_min", "image_max", "image_mean", "image_std", "image_p01", "image_median", "image_p99", "negative_fraction", "low_intensity_fraction", "mask_image_mean", "mask_image_std", "mask_image_min", "mask_image_max", "mask_low_intensity_fraction"]
summary_columns = ["count", "mean", "std", "min", "median", "max"]
geometry_summary = synthetic_geometry.groupby("dataset")[geometry_columns].agg(summary_columns).round(4)
image_summary = synthetic_geometry.groupby("dataset")[image_columns].agg(summary_columns).round(4)

print("Synthetic mask geometry: before and after filtering")
display(geometry_summary)
print("Synthetic image statistics: before and after filtering")
display(image_summary)

print("Sample counts and exclusion rate")
display(pd.DataFrame({
    "Dataset": ["Synthetic all", "Synthetic filtered", "Excluded"],
    "Samples": [len(synthetic_geometry_all), len(synthetic_geometry_filtered), len(synthetic_geometry_all) - len(synthetic_geometry_filtered)],
    "Percentage of all samples": [100.0, round(100 * len(synthetic_geometry_filtered) / len(synthetic_geometry_all), 2), round(100 * (len(synthetic_geometry_all) - len(synthetic_geometry_filtered)) / len(synthetic_geometry_all), 2)]
}))

geometry_summary.to_csv("synthetic_mask_geometry_summary.csv")
image_summary.to_csv("synthetic_image_statistics_summary.csv")
synthetic_geometry.to_csv("synthetic_per_image_statistics.csv", index=False)
with open("synthetic_statistics.txt", "w", encoding="utf-8") as file:
    file.write("Synthetic mask geometry summary\n")
    file.write(geometry_summary.to_string())
    file.write("\n\nSynthetic image statistics summary\n")
    file.write(image_summary.to_string())
print("Saved CSV tables and synthetic_statistics.txt in the notebook working directory.")

plot_columns = ["mask_area_px", "image_mean", "image_std", "low_intensity_fraction", "mask_low_intensity_fraction"]
fig, axes = plt.subplots(1, 5, figsize=(24, 5))
for ax, column in zip(axes.flat, plot_columns):
    for dataset_name, frame in synthetic_geometry.groupby("dataset"):
        values = frame[column].replace([np.inf, -np.inf], np.nan).dropna()
        ax.hist(values, bins=50, alpha=0.55, density=True, label=dataset_name)
    ax.set_title(column.replace("_", " ").title())
    ax.set_ylabel("Density")
    ax.legend()
    ax.grid(alpha=0.25)
plt.suptitle("Synthetic data before and after conservative filtering")
plt.tight_layout()
plt.show()

# Connected components are discrete counts, so use integer-aligned bins.
fig, ax = plt.subplots(figsize=(11, 5))
bins = np.arange(-0.5, 20.5, 1)
for dataset_name, frame in synthetic_geometry.groupby("dataset"):
    values = frame.loc[frame["components_4"] <= 20, "components_4"].dropna()
    ax.hist(values, bins=bins, alpha=0.6, label=dataset_name) # type: ignore
ax.set_xticks(range(0, 20))
ax.set_xlabel("4-connected components (0–19 shown)")
ax.set_ylabel("Number of images")
ax.set_title("Connected-component distribution for the central range")
ax.legend()
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()

# Show the extreme fragmentation tail separately.
fig, ax = plt.subplots(figsize=(11, 5))
for dataset_name, frame in synthetic_geometry.groupby("dataset"):
    values = frame.loc[frame["components_4"] > 10, "components_4"].dropna()
    if not values.empty:
        ax.hist(values, bins=50, alpha=0.6, label=dataset_name) # type: ignore
ax.set_xlabel("4-connected components (>10)")
ax.set_ylabel("Number of images")
ax.set_title("Extreme connected-component tail")
ax.legend()
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

### 2 Visualization

 real and synthetic BRAVO slices together with their segmentation masks. Each selected case is shown as three panels: the BRAVO image, the image with the segmentation contour overlaid, and the segmentation mask. Selection can be random and reproducible, or manually specified by filename or dataset index.

```python
def _remove_modality_suffix(filename, modality):
    """Return a filename stem without NIfTI and modality suffixes."""
    stem = filename.replace(".nii.gz", "").replace(".nii", "")
    for suffix in (f"_{modality}", f".{modality}"):
        if stem.endswith(suffix):
            stem = stem[: -len(suffix)]
    return stem

def collect_bravo_seg_pairs(data_dir, modality="bravo"):
    """Collect and validate BRAVO/mask pairs from nested or flat layouts."""
    data_dir = Path(data_dir)
    image_files = sorted(data_dir.glob(f"**/{modality}/*.nii*"))
    mask_files = sorted(data_dir.glob("**/seg/*.nii*"))

    images = {}
    for image_path in image_files:
        relative = image_path.relative_to(data_dir)
        group = relative.parts[-3] if len(relative.parts) >= 3 else "flat"
        key = (group, _remove_modality_suffix(image_path.name, modality))
        images[key] = image_path

    masks = {}
    for mask_path in mask_files:
        relative = mask_path.relative_to(data_dir)
        group = relative.parts[-3] if len(relative.parts) >= 3 else "flat"
        key = (group, _remove_modality_suffix(mask_path.name, "seg"))
        masks[key] = mask_path

    missing_masks = sorted(set(images) - set(masks))
    missing_images = sorted(set(masks) - set(images))
    if missing_masks or missing_images:
        raise ValueError(
            f"Could not pair BRAVO and segmentation files in {data_dir}. "
            f"Missing masks: {len(missing_masks)}; missing images: {len(missing_images)}."
        )

    return [
        {"image": images[key], "mask": masks[key], "key": key}
        for key in sorted(images)
    ]

def _display_range(image):
    """Choose a stable display range while ignoring empty background."""
    nonzero = image[np.isfinite(image) & (image != 0)]
    if nonzero.size == 0:
        return float(np.nanmin(image)), float(np.nanmax(image))
    return tuple(np.percentile(nonzero, (1, 99)))
```

```python
def _normalize_case_name(name):
    """Reduce a filename or path to the stem used in case keys."""
    stem = _remove_modality_suffix(Path(str(name)).name, "bravo")
    return _remove_modality_suffix(stem, "seg")

def _select_cases_by_name(cases, names):
    stem_to_indices = {}
    for index, case in enumerate(cases):
        stem_to_indices.setdefault(case["key"][1], []).append(index)

    selected_indices = []
    missing = []
    for name in names:
        matches = stem_to_indices.get(_normalize_case_name(name), [])
        if not matches:
            missing.append(name)
        elif len(matches) > 1:
            candidates = [str(cases[index]["image"]) for index in matches]
            raise ValueError(f"Name {name!r} matches several cases: {candidates}")
        else:
            selected_indices.append(matches[0])
    if missing:
        raise KeyError(f"No cases found for names: {missing}")
    return selected_indices

def _select_cases(cases, count, selection, indices, seed, names=None):
    if names is not None:
        selected_indices = _select_cases_by_name(cases, names)
    elif indices is not None:
        selected_indices = list(indices)
        invalid = [index for index in selected_indices if index < 0 or index >= len(cases)]
        if invalid:
            raise IndexError(f"Selected indices out of range: {invalid}")
    elif selection == "random":
        if count < 1:
            raise ValueError("count must be at least 1")
        if count > len(cases):
            raise ValueError(f"Requested {count} cases, but only {len(cases)} are available.")
        rng = random.Random(seed)
        selected_indices = sorted(rng.sample(range(len(cases)), count))
    elif selection == "first":
        selected_indices = list(range(min(count, len(cases))))
    else:
        raise ValueError("selection must be 'random' or 'first' when indices are not provided")

    return [(index, cases[index]) for index in selected_indices]

def _load_case(case):
    image = np.squeeze(nib.load(case["image"]).get_fdata()) # type: ignore
    mask = np.squeeze(nib.load(case["mask"]).get_fdata()) # type: ignore
    if image.ndim != 2 or mask.ndim != 2:
        raise ValueError(
            f"Expected 2D files, got image shape {image.shape} and mask shape {mask.shape} "
            f"for {case['image'].name}."
        )
    if image.shape != mask.shape:
        raise ValueError(
            f"Image/mask shape mismatch for {case['image'].name}: "
            f"{image.shape} versus {mask.shape}."
        )
    return image, mask

def show_dataset_cases(cases, selected_cases, dataset_name, figsize_per_case=(12, 3.6)):
    """Plot BRAVO, contour overlay, and mask for selected cases."""
    rows = len(selected_cases)
    fig, axes = plt.subplots(
        rows,
        3,
        figsize=(figsize_per_case[0], figsize_per_case[1] * rows),
        squeeze=False,
    )
    fig.suptitle(f"{dataset_name} BRAVO and segmentation examples", fontsize=15)

    for row, (index, case) in enumerate(selected_cases):
        image, mask = _load_case(case)
        vmin, vmax = _display_range(image)
        binary_mask = mask > 0

        axes[row, 0].imshow(image, cmap="gray", vmin=vmin, vmax=vmax)
        axes[row, 0].set_title(f"BRAVO | index {index}")

        axes[row, 1].imshow(image, cmap="gray", vmin=vmin, vmax=vmax)
        if binary_mask.any():
            axes[row, 1].contour(binary_mask, levels=[0.5], colors="red", linewidths=1.5)
        axes[row, 1].set_title("BRAVO + mask contour")

        axes[row, 2].imshow(binary_mask, cmap="gray", interpolation="nearest")
        axes[row, 2].set_title("Segmentation mask")

        for column in range(3):
            axes[row, column].axis("off")

        print(f"{dataset_name} index {index}: {case['image']}")
        print(f"{dataset_name} mask {index}:  {case['mask']}")

    plt.tight_layout()
    plt.show()
```

```python
def visualize_real_and_synthetic(
    real_dir,
    synthetic_dir,
    count=3,
    selection="random",
    real_indices=None,
    synthetic_indices=None,
    seed=42,
    real_names=None,
    synthetic_names=None,
):
    """Display selected real and synthetic cases in separate figures.

    Use ``selection='random'`` for reproducible random samples. For manual
    selection, pass ``real_names`` and ``synthetic_names`` with filename stems
    (e.g. ``"Mets_005_slice041"`` or ``"5"``; full BRAVO/seg filenames also
    work), or ``real_indices`` and ``synthetic_indices`` with positions in the
    sorted validated pair lists. Names take precedence over indices, and each
    dataset is resolved independently.
    """
    real_cases = collect_bravo_seg_pairs(real_dir)
    synthetic_cases = collect_bravo_seg_pairs(synthetic_dir)
    print(f"Validated pairs: real={len(real_cases)}, synthetic={len(synthetic_cases)}")

    real_selected = _select_cases(
        real_cases, count, selection, real_indices, seed, real_names
    )
    synthetic_selected = _select_cases(
        synthetic_cases, count, selection, synthetic_indices, seed + 1, synthetic_names
    )

    show_dataset_cases(real_cases, real_selected, "Real")
    show_dataset_cases(synthetic_cases, synthetic_selected, "Synthetic")
```

```python
def _transform_for_visualization(array, transform_mode, target_size, interpolation="bilinear"):
    """Apply only the selected spatial transform to a 2D array."""
    if transform_mode not in {"original", "resize", "pad_crop"}:
        raise ValueError("transform_mode must be 'original', 'resize', or 'pad_crop'")

    array = np.asarray(array)
    if array.ndim != 2:
        raise ValueError(f"Expected a 2D array, got shape {array.shape}")
    if transform_mode == "original":
        return array.copy()

    channel_first = array[None, ...]
    if transform_mode == "resize":
        transformed = Resize(
            spatial_size=target_size,
            mode=interpolation,
        )(channel_first) # type: ignore
    else:
        transformed = ResizeWithPadOrCrop(
            spatial_size=target_size,
            mode="constant",
        )(channel_first) # type: ignore
    return np.asarray(transformed).squeeze(0)

def _visualization_display_range(image, display_range):
    """Return plotting limits without changing the image array."""
    if display_range == "raw":
        return None
    if display_range == "percentile":
        return _display_range(image)
    raise ValueError("display_range must be 'raw' or 'percentile'")
```

```python
def show_spatial_comparison(
    cases,
    selected_cases,
    dataset_name,
    transform_mode="pad_crop",
    target_size=(256, 256),
    display_range="raw",
    figsize_per_case=(16, 3.8),
):
    """Compare stored images with one selected spatial preprocessing mode."""
    rows = len(selected_cases)
    fig, axes = plt.subplots(
        rows,
        4,
        figsize=(figsize_per_case[0], figsize_per_case[1] * rows),
        squeeze=False,
    )
    fig.suptitle(
        f"{dataset_name}: original vs {transform_mode} at {target_size} "
        f"(display range: {display_range})",
        fontsize=15,
    )

    for row, (index, case) in enumerate(selected_cases):
        image, mask = _load_case(case)
        transformed_image = _transform_for_visualization(
            image, transform_mode, target_size, interpolation="bilinear"
        )
        transformed_mask = _transform_for_visualization(
            mask, transform_mode, target_size, interpolation="nearest"
        ) > 0.5
        vmin, vmax = _visualization_display_range(image, display_range) or (None, None)

        axes[row, 0].imshow(image, cmap="gray", vmin=vmin, vmax=vmax, interpolation="nearest")
        axes[row, 0].set_title(f"Original {image.shape}")

        axes[row, 1].imshow(
            transformed_image,
            cmap="gray",
            vmin=vmin,
            vmax=vmax,
            interpolation="nearest",
        )
        axes[row, 1].set_title(f"{transform_mode} {transformed_image.shape}")

        axes[row, 2].imshow(
            transformed_image,
            cmap="gray",
            vmin=vmin,
            vmax=vmax,
            interpolation="nearest",
        )
        if transformed_mask.any():
            axes[row, 2].contour(transformed_mask, levels=[0.5], colors="red", linewidths=1.5)
        axes[row, 2].set_title("Transformed + mask contour")

        axes[row, 3].imshow(transformed_mask, cmap="gray", interpolation="nearest")
        axes[row, 3].set_title("Transformed mask")

        for column in range(4):
            axes[row, column].axis("off")

        print(
            f"{dataset_name} index {index}: original {image.shape} -> "
            f"{transform_mode} {transformed_image.shape}"
        )
        print(f"{dataset_name} image: {case['image']}")
        print(f"{dataset_name} mask:  {case['mask']}")

    plt.tight_layout()
    plt.show()
```

```python
def visualize_spatial_comparison(
    real_dir,
    synthetic_dir,
    count=3,
    selection="random",
    real_indices=None,
    synthetic_indices=None,
    seed=42,
    transform_mode="pad_crop",
    target_size=(256, 256),
    display_range="raw",
    real_names=None,
    synthetic_names=None,
):
    """Visualize real and synthetic cases with an explicit spatial mode."""
    if transform_mode not in {"original", "resize", "pad_crop"}:
        raise ValueError("transform_mode must be 'original', 'resize', or 'pad_crop'")

    real_cases = collect_bravo_seg_pairs(real_dir)
    synthetic_cases = collect_bravo_seg_pairs(synthetic_dir)
    real_selected = _select_cases(
        real_cases, count, selection, real_indices, seed, real_names
    )
    synthetic_selected = _select_cases(
        synthetic_cases, count, selection, synthetic_indices, seed + 1, synthetic_names
    )

    show_spatial_comparison(
        real_cases,
        real_selected,
        "Real",
        transform_mode=transform_mode,
        target_size=target_size,
        display_range=display_range,
    )
    show_spatial_comparison(
        synthetic_cases,
        synthetic_selected,
        "Synthetic",
        transform_mode=transform_mode,
        target_size=target_size,
        display_range=display_range,
    )
```
